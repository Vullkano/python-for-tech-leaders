## **2.1. Caracteres Especiais e Sequências de Escape**

Ao trabalhar com ficheiros de texto é comum lidarmos com **sequências de escape** — combinações de caracteres especiais iniciadas com uma barra invertida `\`.

Estas sequências permitem representar **caracteres invisíveis ou especiais**, como quebras de linha, tabulações, aspas, entre outros.

**Tabela de Sequências de Escape em Python**

| Sequência | Significado                         | Exemplo (String)             | Resultado real                   |
|-----------|-------------------------------------|-------------------------------|----------------------------------|
| `\n`      | Nova linha (*newline*)              | `"linha 1\nlinha 2"`          | linha 1 <br> linha 2             |
| `\t`      | Tabulação horizontal                | `"palavra1\tpalavra2"`        | palavra1    palavra2             |
| `\\`      | Barra invertida literal (`\`)       | `"c:\\ficheiro\\texto.txt"`   | c:\ficheiro\texto.txt            |

Quando trabalhamos com grandes quantidades de texto é comum termos de **ler, escrever e processar ficheiros de texto**.
O python oferece uma forma simples de interagir com ficheiros através da função `open()`.

| Conceito                        | Descrição                                                                 |
|--------------------------------|---------------------------------------------------------------------------|
| `open('ficheiro.txt')`         | Abre um ficheiro em modo de leitura (por omissão).                        |
| `open('ficheiro.txt', 'w')`    | Abre um ficheiro em modo de escrita (apaga conteúdo anterior).            |
| `open('ficheiro.txt', 'a')`    | Abre um ficheiro em modo de adição (escreve no final).                    |
| `open('ficheiro.txt', 'r')`    | Abre um ficheiro em modo de leitura.                                      |
| `read()`                       | Lê todo o conteúdo do ficheiro.                                           |
| `readline()`                   | Lê uma linha do ficheiro.                                                 |
| `readlines()`                  | Lê todas as linhas e devolve uma lista.                                   |
| `write(texto)`                 | Escreve texto no ficheiro.                                                |
| `close()`                      | Fecha o ficheiro. É importante para libertar recursos.                    |
| `with open(...) as f:`         | Abre o ficheiro de forma segura (fecha automaticamente após uso).         |

**Exemplo 1: Ler um ficheiro linha a linha**

````python
x = open('text.txt', 'r', encoding='utf-8')
for i in x:
    print(i.strip())
````

### **3.1.1. Barplot**
- Abaixo apresenta-se uma função simples para gerar um gráfico de barras (barplot). Iremos visualizar a frequência de palavras contidas no ficheiro, seguindo-se a representação das 10 palavras mais frequentes.

````python
def plot_barplot(abcissas, odenadas, titulo='Palavras mais frequentes', horizontal=False):
    plt.figure(figsize=(10, 5))
    if horizontal:
        plt.barh(list(abcissas), list(odenadas), color='skyblue', edgecolor='black')
        plt.xlabel('Frequência')
        plt.ylabel('Palavras')
    else:
        plt.bar(list(abcissas), list(odenadas), color='skyblue', edgecolor='black')
        plt.xlabel('Palavras')
        plt.ylabel('Frequência')
        plt.xticks(rotation=90)
    plt.title(titulo, fontsize=14, weight='bold')
    plt.tight_layout()
    plt.show()
````

**a) Visualizar a frequência de palavras**

````python
plot_barplot(dict_contar.keys(), dict_contar.values())
````

**b) Visualizar Top10 palavras mais frequentes**

````python
high_contar = {}

for palavra, frequencia in dict_contar.items():
    if frequencia >= 2:
        high_contar[palavra] = frequencia

plot_barplot(list(high_contar.keys()), list(high_contar.values()), titulo='Top10 Palavras Mais Frequentes', horizontal=True)
````

### **3.1.2. Lineplot**
- O lineplot (gráfico de linhas) é uma representação visual que mostra a evolução ou variação de um valor ao longo de um eixo, normalmente o tempo ou a posição.
- Neste caso, vamos utilizar um lineplot para visualizar o número de caracteres por linha no ficheiro de texto. Esta abordagem ajuda a perceber como varia o comprimento das linhas ao longo da obra, o que pode ser relevante para identificar estruturas formais, mudanças de estilo ou seções distintas.

````python
def plot_lineplot(abcissas, odenadas, titulo='Contagem de Linhas', xlabel='Eixo X', ylabel='Eixo Y'):
    # Criar o gráfico de linha
    plt.figure(figsize=(10, 5))
    plt.plot(list(abcissas), list(odenadas), marker='o', linestyle='-', color='royalblue', linewidth=2, markersize=6)

    # Personalização do gráfico
    plt.title(titulo, fontsize=14, weight='bold')
    plt.xlabel(xlabel, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
````

**a) Visualiza o Tamanho das frases ao longo das linhas**

````python
alinea_count = {}
with open('text.txt', 'r', encoding='utf-8') as file:
    for number, linha in enumerate(file):
        print(f"Linha {number}: {len(linha.strip())}")
        alinea_count[number] = len(linha.strip())
````

````python
plot_lineplot(alinea_count.keys(), alinea_count.values(), 
              titulo='Comprimento das Linhas (Caracteres)', 
              xlabel='Número da Linha', 
              ylabel='Número de Caracteres')
````

### **3.1.3. Pie Chart**
- O pie chart (gráfico circular) é uma representação gráfica usada para mostrar a proporção de diferentes categorias dentro de um todo.
- Vamos utilizá-lo para visualizar a distribuição das vogais no ficheiro de texto, permitindo perceber quais são mais frequentes. Esta análise pode ser útil em estudos linguísticos e fonológicos dentro do contexto da filologia digital.

````python
def plot_piechart(labels, valores, titulo='Distribuição de categorias'):
    """Cria um gráfico circular a partir de categorias e respetivos valores."""
    plt.figure(figsize=(7, 7))
    plt.pie(list(valores), labels=list(labels), autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)

    plt.title(titulo, fontsize=14, weight='bold')
    plt.axis('equal')  # Garante que o círculo fique redondo
    plt.tight_layout()
    plt.show()
````

**a) Visualiza a distribuição das vogais ao longo do ficheiro**

````python
vogal_count = {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}
with open('text.txt', 'r', encoding='utf-8') as file:
    for linha in file:
        for char in linha.lower():
            if char in vogal_count:
                vogal_count[char] += 1

for i, j in vogal_count.items():
    print(f"{i}: {j}")
````

````python
plot_piechart(vogal_count.keys(), vogal_count.values(), titulo='Distribuição de Vogais no Ficheiro')
````

### **3.1.4. Scatter Plot**
- O scatter plot (gráfico de dispersão) é uma visualização que mostra a relação entre duas variáveis numéricas.
- Neste contexto, será utilizado para representar a relação entre o número de palavras e o número de caracteres por linha do ficheiro analisado.
- Este tipo de gráfico é útil para perceber, por exemplo, se as linhas mais longas (em caracteres) também tendem a ter mais palavras, ou se existem linhas curtas mas densas em termos linguísticos.

````python
def plot_scatter(x, y, titulo='Relação entre variáveis', xlabel='Variável X', ylabel='Variável Y'):
    """Cria um gráfico de dispersão (scatter plot) a partir de duas listas de valores."""
    plt.figure(figsize=(10, 5))
    plt.scatter(list(x), list(y), color='teal', alpha=0.8, edgecolors='black', s=80)

    plt.title(titulo, fontsize=14, weight='bold')
    plt.xlabel(xlabel, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
````

**a) Visualiza a relação entre o comprimento das palavras e a sua frequência de aparição**

````python
comp_freq = {}

with open('text.txt', 'r', encoding='utf-8') as file:
    for linha in file:
        palavras = linha.strip().split()
        palavras = [palavra.strip(",.;") for palavra in palavras]
        for palavra in palavras:
            if len(palavra) in comp_freq.keys():
                comp_freq[len(palavra)] += 1
            else:
                comp_freq[len(palavra)] = 1

for i, j in comp_freq.items():
    print(f"Comprimento {i}: {j} aparições")
````

````python
plot_scatter(comp_freq.keys(), comp_freq.values(),
             titulo='Comprimento das palavras vs. frequência',
             xlabel='Comprimento da palavra',
             ylabel='Frequência')
````

### **3.2.1. Barplot**
- Abaixo apresenta-se uma função simples para gerar um gráfico de barras (barplot). Iremos visualizar a frequência de palavras contidas no ficheiro, seguindo-se a representação das 10 palavras mais frequentes.

````python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_barplot_sns(x, y, titulo='Frequência de Palavras', horizontal=False):
    # 1. Configuração do tema com a fonte padrão do sistema e eixos suaves
    sns.set_theme(style="white", rc={
        "axes.edgecolor": "#cbd5e1"
    })
    
    plt.figure(figsize=(10, 5))

    data = sorted(zip(x, y), key=lambda item: item[1], reverse=True)
    palavras, frequencias = zip(*data)

    # 2. Palete de cores sequencial sofisticada (mako)
    cores = sns.color_palette("mako_r", n_colors=len(palavras) + 4)[2:len(palavras)+2]

    # 3. Desenho dos gráficos
    if horizontal:
        ax = sns.barplot(
            x=list(frequencias), 
            y=list(palavras), 
            hue=list(palavras), 
            legend=False, 
            palette=cores, 
            edgecolor="none",          
            linewidth=0
        )
            
        plt.xlabel('Frequência', fontweight='bold', color='#475569', labelpad=10)
        plt.ylabel('Palavras', fontweight='bold', color='#475569', labelpad=10)
        
        # NUMERAÇÃO INFALÍVEL PARA HORIZONTAL: Itera sobre cada barra desenhada
        for patch in ax.patches:
            width = patch.get_width()
            if width > 0: # Garante que só numera barras com valor
                ax.text(
                    width + (max(frequencias) * 0.015), # Posição X (um pouco à frente da barra)
                    patch.get_y() + patch.get_height() / 2, # Posição Y (centro vertical da barra)
                    f'{int(width)}', 
                    ha='left', 
                    va='center', 
                    weight='bold', 
                    color='#334155', 
                    fontsize=10
                )
        
        # Garante espaço à direita para os números não serem cortados
        ax.set_xlim(0, max(frequencias) * 1.15)
    else:
        ax = sns.barplot(
            x=list(palavras), 
            y=list(frequencias), 
            hue=list(palavras), 
            legend=False, 
            palette=cores, 
            edgecolor="none",          
            linewidth=0
        )
            
        plt.xlabel('Palavras', fontweight='bold', color='#475569', labelpad=10)
        plt.ylabel('Frequência', fontweight='bold', color='#475569', labelpad=10)
        plt.xticks(rotation=45, color='#334155')
        
        # NUMERAÇÃO INFALÍVEL PARA VERTICAL: Itera sobre cada barra desenhada
        for patch in ax.patches:
            height = patch.get_height()
            if height > 0: # Garante que só numera barras com valor
                ax.text(
                    patch.get_x() + patch.get_width() / 2, # Posição X (centro horizontal da barra)
                    height + (max(frequencias) * 0.015), # Posição Y (um pouco acima da barra)
                    f'{int(height)}', 
                    ha='center', 
                    va='bottom', 
                    weight='bold', 
                    color='#334155', 
                    fontsize=10
                )
        
        # Garante espaço no topo para os números não serem cortados
        ax.set_ylim(0, max(frequencias) * 1.15)

    # 4. Título e polimento final dos eixos
    ax.set_title(titulo, fontsize=15, weight='bold', pad=20, color='#0f172a')
    ax.tick_params(colors='#64748b')   
    
    sns.despine(left=True if horizontal else False, bottom=False if horizontal else True)
    plt.tight_layout()
    plt.show()

# Chamada da função (Exemplo de uso)
plot_barplot_sns(high_contar.keys(), high_contar.values(), titulo='Top 10 Palavras Mais Frequentes', horizontal=True)
````

- Neste caso, vamos utilizar um lineplot para visualizar o número de caracteres por linha no ficheiro de texto. Esta abordagem ajuda a perceber como varia o comprimento das linhas ao longo da obra, o que pode ser relevante para identificar estruturas formais, mudanças de estilo ou seções distintas.

````python
def plot_lineplot_sns(x, y, titulo='Número de Caracteres por Linha'):
    # 1. Configuração de um tema minimalista e moderno
    sns.set_theme(style="whitegrid", rc={
        "grid.color": "#f1f5f9",
        "axes.edgecolor": "#e2e8f0",
    })
    
    plt.figure(figsize=(12, 5.5)) # Ligeiramente mais alto para dar espaço aos textos do topo
    
    # 2. Desenho da linha principal com cores ligeiramente ajustadas para maior harmonia
    ax = sns.lineplot(
        x=list(x), y=list(y), 
        marker='o', 
        markersize=9, 
        linewidth=3, 
        color='#0284c7',       # Azul ligeiramente mais profundo e profissional
        markerfacecolor='white',
        markeredgecolor='#0284c7',
        markeredgewidth=2.5
    )
    
    # 3. Adiciona um preenchimento (sombra) gradiente suave abaixo da linha
    plt.fill_between(list(x), list(y), color='#0ea5e9', alpha=0.08)
    
    # 4. Posicionamento dinâmico dos valores acima dos pontos
    max_y = max(y) if y else 1
    offset = max_y * 0.025     # Calcula a distância ideal com base no maior valor
    
    for xi, yi in zip(x, y):
        ax.text(
            xi, 
            yi + offset, 
            f'{yi:,}',          # Formata com separador de milhares se necessário
            ha='center', 
            va='bottom', 
            fontsize=9.5, 
            weight='semibold', 
            color='#1e293b'     # Slate escuro em vez de preto puro
        )
    
    # 5. Ajustes finais de eixos, títulos e limites
    ax.set_title(titulo, fontsize=15, weight='bold', pad=20, color='#0f172a')
    ax.set_xlabel('Número da Linha', fontsize=11, fontweight='bold', color='#334155', labelpad=10)
    ax.set_ylabel('Número de Caracteres', fontsize=11, fontweight='bold', color='#334155', labelpad=10)
    ax.set_xticks(list(x))
    
    # Garante que os rótulos do topo não são cortados pelo limite do gráfico
    ax.set_ylim(min(y) - offset*2 if y else 0, max_y + offset * 4)
    
    sns.despine(left=True, bottom=True)
    plt.tight_layout()
    plt.show()

# Chamada da função
plot_lineplot_sns(alinea_count.keys(), alinea_count.values(), titulo='Contagem de Caracteres por Linha')
````

### **3.2.3. Pie Chart**
- O pie chart (gráfico circular) é uma representação gráfica usada para mostrar a proporção de diferentes categorias dentro de um todo.
- Vamos utilizá-lo para visualizar a distribuição das vogais no ficheiro de texto, permitindo perceber quais são mais frequentes. Esta análise pode ser útil em estudos linguísticos e fonológicos dentro do contexto da filologia digital.

````python
def plot_piechart_beautiful(labels, valores, titulo='Distribuição de categorias'):
    total = sum(valores)
    
    def func_format(pct):
        val_absoluto = int(round(pct * total / 100.0))
        return f'{pct:.1f}%\n({val_absoluto})'

    plt.figure(figsize=(7, 7))
    
    cores = sns.color_palette('muted')[0:len(labels)]
    explode = [0.05 if v == max(valores) else 0 for v in valores]
    
    wedges, texts, autotexts = plt.pie(
        list(valores),
        labels=list(labels),
        autopct=func_format,
        startangle=90,
        colors=cores,
        explode=explode,
        shadow=False,
        pctdistance=0.75,
        textprops={'fontsize': 11, 'weight': 'bold', 'color': '#1e293b'},
        wedgeprops={'linewidth': 2, 'edgecolor': 'white'}
    )
    
    circulo_central = plt.Circle((0,0), 0.60, fc='white')
    fig = plt.gcf()
    fig.gca().add_artist(circulo_central)
    
    for autotext in autotexts:
        autotext.set_color('white')
        autotext.set_fontsize(10)
        autotext.set_linespacing(1.2)
    
    for text in texts:
        text.set_color('#0f172a')
        text.set_fontsize(12)
        text.set_weight('bold')
        
    plt.title(titulo, fontsize=14, weight='bold', pad=15)
    plt.axis('equal')
    plt.tight_layout()
    plt.show()

plot_piechart_beautiful(vogal_count.keys(), vogal_count.values(), titulo='Distribuição de Vogais no Ficheiro')
````

### **3.2.4. Scatter Plot**
- O scatter plot (gráfico de dispersão) é uma visualização que mostra a relação entre duas variáveis numéricas.
- Neste contexto, será utilizado para representar a relação entre o número de palavras e o número de caracteres por linha do ficheiro analisado.
- Este tipo de gráfico é útil para perceber, por exemplo, se as linhas mais longas (em caracteres) também tendem a ter mais palavras, ou se existem linhas curtas mas densas em termos linguísticos.

````python
def plot_scatter_sns(x, y, titulo='Relação entre variáveis', xlabel='Variável X', ylabel='Variável Y'):
    # 1. Configuração de um tema limpo e moderno com grelha suave
    sns.set_theme(style="whitegrid", rc={
        "grid.color": "#f1f5f9",
        "axes.edgecolor": "#e2e8f0",
    })
    
    plt.figure(figsize=(10, 6)) # Ligeiramente mais alto para dar espaço aos rótulos
    
    # 2. Scatter plot com mapeamento de cor (gradiente) baseado nos valores de Y
    ax = sns.scatterplot(
        x=list(x),
        y=list(y),
        hue=list(y),
        palette="ch:start=.2,rot=-.3", # Gradiente elegante de azul/escuro
        s=160,                         # Pontos ligeiramente maiores
        edgecolor='white',             # Contorno branco para destacar fatias sobrepostas
        linewidth=1.5,
        alpha=0.95,
        legend=False                   # Remove a legenda automática já que a cor segue o eixo Y
    )

    # 3. Posicionamento inteligente dos rótulos de texto
    max_y = max(y) if y else 1
    offset = max_y * 0.025            # Calcula o espaço vertical ideal com base no maior valor
    
    for xi, yi in zip(x, y):
        ax.text(
            xi, 
            yi + offset, 
            f'{yi:,}',                 # Formata números grandes com separador de milhares, se aplicável
            ha='center', 
            va='bottom', 
            fontsize=9.5, 
            weight='semibold', 
            color='#475569'            # Cinza escuro suave, menos agressivo que preto puro
        )

    # 4. Tipografia e polimento dos eixos
    ax.set_title(titulo, fontsize=15, weight='bold', pad=20, color='#0f172a')
    ax.set_xlabel(xlabel, fontsize=11, fontweight='bold', color='#334155', labelpad=10)
    ax.set_ylabel(ylabel, fontsize=11, fontweight='bold', color='#334155', labelpad=10)
    
    # Ajusta os limites do eixo Y para garantir que os rótulos do topo não fiquem cortados
    ax.set_ylim(min(y) - offset*2 if y else 0, max_y + offset * 4)
    
    sns.despine(left=True, bottom=True)
    plt.tight_layout()
    plt.show()

# Chamada da função
plot_scatter_sns(comp_freq.keys(), comp_freq.values(),
                 titulo='Comprimento das Palavras vs. Frequência',
                 xlabel='Comprimento da Palavra',
                 ylabel='Frequência')
````
